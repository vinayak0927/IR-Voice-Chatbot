```python
import re
from gtts import gTTS
from IPython.display import Audio, display

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# ==========================================
# KNOWLEDGE BASE
# ==========================================

knowledge_base = [
    {
        "question": "what is python",
        "answer": "Python is a high-level programming language used for web development, data science, artificial intelligence and machine learning."
    },
    {
        "question": "what is machine learning",
        "answer": "Machine learning is a branch of artificial intelligence that enables computers to learn patterns from data and make predictions."
    },
    {
        "question": "what is artificial intelligence",
        "answer": "Artificial intelligence is a field of computer science that focuses on creating systems capable of performing tasks that normally require human intelligence."
    },
    {
        "question": "what is deep learning",
        "answer": "Deep learning is a subset of machine learning that uses neural networks with multiple layers to learn complex patterns."
    },
    {
        "question": "what is computer vision",
        "answer": "Computer vision is a field of artificial intelligence that enables computers to understand images and videos."
    },
    {
        "question": "what is natural language processing",
        "answer": "Natural language processing enables computers to understand, process and generate human language."
    },
    {
        "question": "what are technical skills",
        "answer": "Technical skills are specialized abilities such as programming, database management, machine learning, web development and cybersecurity."
    },
    {
        "question": "what are soft skills",
        "answer": "Soft skills are interpersonal abilities such as communication, teamwork, leadership, problem solving and time management."
    },
    {
        "question": "what is communication skill",
        "answer": "Communication skill is the ability to clearly exchange information, ideas and opinions with other people."
    },
    {
        "question": "what is teamwork",
        "answer": "Teamwork is the ability to work effectively with other people to achieve a common goal."
    }
]
```

```python
# ==========================================
# TEXT CLEANING
# ==========================================

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)
    return text


questions = [
    clean_text(item["question"])
    for item in knowledge_base
]
```

```python
# ==========================================
# TF-IDF
# ==========================================

vectorizer = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2)
)

question_vectors = vectorizer.fit_transform(questions)
```

```python

# ==========================================
# FIND ANSWER
# ==========================================

def find_answer(user_question):

    user_question = clean_text(user_question)

    user_vector = vectorizer.transform([user_question])

    similarity_scores = cosine_similarity(
        user_vector,
        question_vectors
    )[0]

    best_index = similarity_scores.argmax()

    best_score = similarity_scores[best_index]

    if best_score < 0.20:
        return (
            "Sorry, I could not find a suitable answer "
            "for your question.",
            best_score
        )

    return (
        knowledge_base[best_index]["answer"],
        best_score
    )

```

```python
# ==========================================
# TEXT TO SPEECH
# ==========================================

def speak(text):

    tts = gTTS(
        text=text,
        lang="en"
    )

    filename = "answer.mp3"

    tts.save(filename)

    display(Audio(filename, autoplay=True))
```

```python



# ==========================================
# CHATBOT
# ==========================================

def chatbot():

    print("=" * 60)
    print("          IR BASED VOICE CHATBOT")
    print("=" * 60)

    while True:

        user_question = input("\nYou: ")

        if user_question.lower().strip() in [
            "exit",
            "quit",
            "bye"
        ]:

            answer = "Goodbye! Have a nice day."

            print("Bot:", answer)

            speak(answer)

            break

        if not user_question.strip():
            continue

        answer, score = find_answer(user_question)

        print("\nBot:", answer)

        print(
            "Similarity Score:",
            round(score, 3)
        )

        # Speak answer
        speak(answer)
```

```python
# ==========================================
# START
# ==========================================

chatbot()
```
